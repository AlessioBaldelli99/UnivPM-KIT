# Use ldpc package to compute X, Z logical operators 

```python
import numpy as np
from bposd.css import css_code
import galois

import sys
from convert_alist import *
```

```python
# sys.path.append('./alists/')

# n, k, d = 64, 18, 8
# n, k, d = 65, 27, 4
# n, k, d = 128, 36, 8
n, k, d = 1025, 583, 33



F2 = galois.GF(2)
alistDirPath = "./alists/PCMs/" 
alistEnd = ".alist"

alistFilePath = alistDirPath + "n" + str(n) + "_k" + str(k) + "_d" + str(d) + "_Hx" + alistEnd
Hx = F2(readAlist(alistFilePath))

alistFilePath = alistDirPath + "n" + str(n) + "_k" + str(k) + "_d" + str(d) + "_Hz" + alistEnd
Hz = F2(readAlist(alistFilePath))
# print("Number of X checks:", Hx.shape[0], "\nNumber of Z checks:", Hz.shape[0])
print("CSS code [[n=%s, k=%s, d=%s]]" % (n, k, d))

comm = Hx @ Hz.T
if np.all(comm == 0):
    print("  X and Z checks commute.")
else:
    print("  X and Z checks do NOT commute.")


# Using bposd.css to construct the CSS code
qcode = css_code(Hx, Hz)
print('Testing CSS code...')
qcode.test()
print('Done')

# lx, lz = qcode.compute_logicals()
lz = qcode.lz.toarray()
lx = qcode.lx.toarray()
print(lx.shape)
print(lz.shape)
```

```python
## Write Lx
alistDirPath_Logicals = "./alists/logicals/"
alistFilePath_Lx = alistDirPath_Logicals + "n" + str(n) + "_k" + str(k) + "_d" + str(d) + "_Lx.alist"
# M = np.array(lx)
out = numpyToAlist(lx)
with open(alistFilePath_Lx, "w") as f:
    f.write(out)

## Write Lz
alistFilePath_Lz = alistDirPath_Logicals + "n" + str(n) + "_k" + str(k) + "_d" + str(d) + "_Lz.alist"
# M = np.array(lz)
out = numpyToAlist(lz)
with open(alistFilePath_Lz, "w") as f:
    f.write(out)
```
